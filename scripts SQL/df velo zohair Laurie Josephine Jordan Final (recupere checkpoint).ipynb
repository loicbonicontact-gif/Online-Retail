## - Comment le nombre de locations de vélos évolue-t-il au cours du temps ?

```python
# Créer le nom du mois (nécessaire pour le groupby et les graphiques)
mois_fr = {1:"Janvier",2:"Février",3:"Mars",4:"Avril",5:"Mai",6:"Juin",
           7:"Juillet",8:"Août",9:"Septembre",10:"Octobre",11:"Novembre",12:"Décembre"}
df_active["month_name"] = df_active["month"].map(mois_fr)

monthly_rentals = (df_active.groupby(["month", "month_name"])["rented_bike_count"].sum().reset_index().sort_values("month"))
```

# <u>Graphique recommandé pour analyse d'une évolution dans le temps c'est une courbe.</u>

```python
fig = px.line(monthly_rentals, x="month_name", y="rented_bike_count", markers=True, title="Évolution mensuelle du nombre de locations de vélos",
              labels = {"month_name": "Mois",
                        "rented_bike_count": "Nombre de locations"})
fig.update_traces(
    line=dict(color="#0d6efd", width=4),
    marker=dict(
        color="#ff7f0e",
        size=9,
        line=dict(color="black",
                  width=1)),textposition="top center")

fig.update_layout(
    template="plotly_white",
    title={
        "text":"📈 Évolution mensuelle du nombre de locations de vélos",
        "x": 0.5},
    xaxis_title="Mois",
    yaxis_title="Nombre de locations",
    hovermode="x unified",
    font=dict(size=14),
    height=500)
                        
fig.show()
```

## La demande augmente progressivement entre le printemps et l'été, atteint un maximum pendant les mois les plus chauds, puis diminue à l'approche de l'hiver.

```python
# Construction du tableau jours ouvrés vs week-end
weekday_vs_weekend = (
    df_active.assign(**{"Type de jour": df_active["is_weekend"].map({True: "Week-end", False: "Jour ouvré"})})
    .groupby("Type de jour", as_index=False)["rented_bike_count"].mean()
)
weekday_vs_weekend
```

```python
# 1. Préparation et calculs avancés des données
df = weekday_vs_weekend.copy()
total_loc = df["rented_bike_count"].sum()
df["part_pct"] = (df["rented_bike_count"] / total_loc) * 100

valeurs = df.set_index("Type de jour")["rented_bike_count"]
jour_max = valeurs.idxmax()
jour_min = valeurs.idxmin()
val_max = valeurs.max()
val_min = valeurs.min()

ecart_abs = val_max - val_min
ecart_pct = ((val_max - val_min) / val_min) * 100

# Calcul de l'écart relatif sur chaque barre
df["delta_vs_min"] = df["rented_bike_count"].apply(
    lambda x: f"+{ecart_pct:.1f}% vs {jour_min}"
    if x == val_max
    else "Base de comparaison"
)

# 2. Palette dynamique : couleur vive pour le leader, plus neutre pour le second
palette = {
    jour_max: "#2b5c8f",  # Bleu profond professionnel
    jour_min: "#8da0cb",}  # Bleu adouci / secondaire


# 3. Création du graphique de base
fig = px.bar(
    df,
    x="Type de jour",
    y="rented_bike_count",
    color="Type de jour",
    color_discrete_map=palette,
    custom_data=["part_pct", "delta_vs_min"],
    title="<b>Comparaison de la demande : Jours ouvrés vs Week-end</b>",
)

# 4. Enrichissement de l'interactivité au survol (Hover Custom)
fig.update_traces(
    texttemplate="<b>%{y:,.0f}</b>",
    textposition="outside",
    textfont=dict(size=14, color="#212529"),
    width=0.45,
    hovertemplate=(
        "<b>Catégorie :</b> %{x}<br>"
        "<b>Moyenne quotidienne :</b> %{y:,.0f} locations<br>"
        "<b>Part du volume total :</b> %{customdata[0]:.1f}%<br>"
        "<b>Écart :</b> %{customdata[1]}<extra></extra>"
    ),
)

# 5. Annotation style "Badge Executive" (Bannière d'accroche)
fig.add_annotation(
    x=0.5,
    y=val_max * 1.22,
    xref="paper",
    yref="y",
    text=f"🔥 <b>{jour_max}</b> surpasse <b>{jour_min}</b> de <b>+{ecart_pct:.0f}%</b> (+{ecart_abs:,.0f} locations/jour)",
    showarrow=False,
    font=dict(size=13, color="#1e293b"),
    bgcolor="#f1f5f9",
    bordercolor="#cbd5e1",
    borderwidth=1,
    borderpad=8,
)

# 6. Ligne visuelle de comparaison (Connecteur d'écart)
fig.add_shape(
    type="line",
    x0=0,
    y0=val_min,
    x1=1,
    y1=val_min,
    line=dict(color="#94a3b8", width=1.5, dash="dot"),
)

# 7. Menu interactif (Boutons pour changer de vue : Valeurs vs Parts %)
fig.update_layout(
    updatemenus=[
        dict(
            type="buttons",
            direction="left",
            x=1.0,
            y=1.18,
            xanchor="right",
            yanchor="top",
            showactive=True,
            buttons=[
                dict(
                    label="Volume Moyen",
                    method="update",
                    args=[
                        {"y": [df["rented_bike_count"]], "texttemplate": "<b>%{y:,.0f}</b>"},
                        {"yaxis": {"title": "Nombre moyen de locations par jour", "range": [0, val_max * 1.3]}},
                    ],
                ),
                dict(
                    label="Part de Marché (%)",
                    method="update",
                    args=[
                        {"y": [df["part_pct"]], "texttemplate": "<b>%{y:.1f}%</b>"},
                        {"yaxis": {"title": "Pourcentage du volume total (%)", "range": [0, 100]}},
                    ],
                ),
            ],
        )
    ],
    template="plotly_white",
    title_x=0.0,  # Titre aligné à gauche (norme moderne de Dataviz)
    title_font=dict(size=18, color="#0f172a", family="Inter, Arial"),
    showlegend=False,
    hoverlabel=dict(
        bgcolor="#ffffff",
        font_size=13,
        font_family="Arial",
        bordercolor="#cbd5e1",
    ),
    xaxis=dict(
        title="",
        showgrid=False,
        tickfont=dict(size=14, color="#334155"),
    ),
    yaxis=dict(
        title="Nombre moyen de locations par jour",
        showgrid=True,
        gridcolor="#f1f5f9",
        range=[0, val_max * 1.3],
    ),
    margin=dict(l=50, r=50, t=110, b=40),
    plot_bgcolor="white",
    paper_bgcolor="white",
)

# 8. Affichage avec barre d'outils nettoyée
fig.show(config={"displayModeBar": True, "displaylogo": False})
```

# Commentaire :
##  On observe que : le nombre moyen de locations est plus élevé les jours ouvrés que durant le week-end ; l'écart est d'environ 1 624 locations supplémentaires par jour ; cela représente une augmentation d'environ 10 % par rapport au week-end.

## Comment la demande évolue-t-elle saison par saison ?

```python
# Construction du tableau des locations moyennes par saison
season_rentals = df_active.groupby("season", as_index=False)["rented_bike_count"].mean()
season_rentals
```

```python
# Identifier la saison ayant la demande moyenne la plus élevée
best_season = season_rentals.loc[
    season_rentals["rented_bike_count"].idxmax()]

# Création de la courbe
fig = px.line(
    season_rentals,
    x="season",
    y="rented_bike_count",
    markers=True,
    text="rented_bike_count",
    title="<b>Évolution de la demande moyenne de locations selon les saisons</b>",
    labels={
        "season": "Saison",
        "rented_bike_count": "Nombre moyen de locations par jour"})

# Personnalisation de la courbe
fig.update_traces(
    line=dict(color="#0d6efd", width=4),
    marker=dict(
        size=10,
        color="#0d6efd",
        line=dict(color="white", width=2)
    ),
    texttemplate="%{text:,.0f}",
    textposition="top center",
    hovertemplate=(
        "<b>Saison :</b> %{x}<br>"
        "<b>Locations moyennes :</b> %{y:,.0f}<extra></extra>"))

# Annotation automatique sur la meilleure saison
fig.add_annotation(
    x=best_season["season"],
    y=best_season["rented_bike_count"],
    text="Demande maximale",
    showarrow=True,
    arrowhead=2,
    arrowsize=1,
    arrowwidth=2,
    arrowcolor="#dc3545",
    ax=0,
    ay=-45,
    bgcolor="#fff3cd",
    bordercolor="#dc3545",
    borderwidth=1,
    font=dict(size=12, color="#212529")
)

# Mise en page
fig.update_layout(
    template="plotly_white",

    title_x=0.5,
    title_font=dict(
        size=18,
        color="#212529"
    ),

    xaxis=dict(
        title="Saison",
        showgrid=False
    ),

    yaxis=dict(
        title="Nombre moyen de locations par jour",
        showgrid=True,
        gridcolor="#f1f5f9"
    ),

    hoverlabel=dict(
        bgcolor="white",
        font_size=13,
        font_family="Arial"
    ),

    font=dict(size=13),

    margin=dict(
        l=50,
        r=40,
        t=80,
        b=40
    ),

    height=500
)

fig.show()
```

## - Comment le nombre de trajets évolue-t-il mois après mois ?

```python
# 1. Conversion de la date et groupement par mois
df_active['YearMonth'] = df_active['date'].dt.to_period('M')

monthly_trips = df_active.groupby('YearMonth')['rented_bike_count'].sum().reset_index()
monthly_trips['YearMonth_str'] = monthly_trips['YearMonth'].astype(str)

# 2. Création du graphique
fig, ax = plt.subplots(figsize=(10, 5))

# Tracé de la courbe
ax.plot(monthly_trips['YearMonth_str'], monthly_trips['rented_bike_count'], 
        color='#d9534f', marker='o', linewidth=2.5, markersize=8, zorder=3)

# Formater l'axe Y en milliers (ex: 800 k)
ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda x, pos: f'{x*1e-3:,.0f} k'))

# Ajouter la valeur au-dessus de chaque point
for i, txt in enumerate(monthly_trips['rented_bike_count']):
    ax.annotate(f'{txt*1e-3:,.0f}k', 
                (monthly_trips['YearMonth_str'][i], monthly_trips['rented_bike_count'][i]),
                textcoords="offset points", 
                xytext=(0, 10), 
                ha='center', fontsize=9, fontweight='bold', color='#333333')

# Ajuster les limites de l'axe Y
ax.set_ylim(0, monthly_trips['rented_bike_count'].max() * 1.15)

# Titres et axes
ax.set_title('Évolution du nombre de trajets mois après mois', fontsize=13, fontweight='bold', pad=15)
ax.set_xlabel('Mois', fontsize=11)
ax.set_ylabel('Nombre total de trajets', fontsize=11)
ax.grid(True, linestyle='--', alpha=0.5, zorder=0)

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Commentaire :
### On peut constater une augmentation du nombre de trajet par moi qui augmente la première moitié de l'année et ensuite diminue de plus en plus.
### La courbe jour par jour montre une grande variabilité (pics et chutes brusques), principalement 
### liée aux épisodes météo défavorables (pluie, neige) sur certains jours précis.

### Effet sur la location de vélos

```python
df_humidity_final = pd.DataFrame({
    'Température moyenne': [
        df_active['temperature'].mean(),
        df_active_negative['temperature'].mean(),
        df_active_positive['temperature'].mean()],
    'Humidité moyenne': [
        df_active['humidity'].mean(),
        df_active_negative['humidity'].mean(),
        df_active_positive['humidity'].mean()],
    'Locations moyennes': [
        df_active['rented_bike_count'].mean(),
        df_active_negative['rented_bike_count'].mean(),
        df_active_positive['rented_bike_count'].mean()]}, index=['Moyenne globale', 'Température < 0°C', 'Température ≥ 0°C'])

df_humidity_final
```

#### **commentaire d'analyse** : on comprends qu'en moyenne, 729 vélos sont utilisés quotidiennement lorsque la température avoisine les 12°c ; la différence la plus flagrante réside entre les températures basses et élevées, avec une différence en moyenne de plus de 639 vélos utilisé quand la température est "élevée" (ici en moyenne 16.51°c).
#### L'explication externe la plus plausible est la prise en compte des températures qui influence l'humeur de l'utilisateur et donc son envie ou non d'emprunter un vélo en libre-service. Il est aussi plus propice qu'un utilisateur souhaite profiter des saisons douces et chaudes en faisant une balade à vélo que pendant la saison hivernale

## - Existe-t-il une température seuil (au-delà/en-deçu) à de laquelle la demande chute ?

### Courbe de tendance :  Suivre l'évolution d'une valeur dans le temps et repérer une tendance, une saisonnalité ou une rupture 

```python
import plotly.express as px

fig = px.scatter(
    df_active,
    x="temperature",
    y="rented_bike_count",
    opacity=0.1,
    trendline="lowess",
    labels={
        "temperature": "Température (°C)",
        "rented_bike_count": "Nombre de vélos loués"
    },
    title="Relation entre température et demande de vélos"
)

fig.update_traces(marker=dict(size=4))

# Cibler uniquement la courbe de tendance pour changer sa couleur
fig.update_traces(
    line=dict(color="red", width=3),
    selector=dict(mode="lines")
)

fig.show()
```

### Commentaire : 
#### Le nuage de points montre une tendance claire : plus la température augmente, plus la demande de vélos augmente. 
#### Mais la courbe de tendance (en rouge) révèle que cette relation n’est pas parfaitement linéaire, et qu’elle se décompose en trois phases distinctes :
#### 1.	En dessous de 0°C : la demande reste très faible et proche de zéro, quasi plate. Le froid semble décourager fortement l’usage du vélo.
#### 2.	Entre 0°C et 25°C environ : la courbe grimpe de façon soutenue, c’est la zone où chaque degré supplémentaire fait le plus gagner en demande.
#### 3.	Au-delà de 25°C : la courbe s’aplatit nettement. La demande continue de progresser très légèrement, mais beaucoup plus lentement comme si un plafond de confort était atteint.

# - La pluie ou la neige ont-elles un effet plus fort que la température sur la baisse de la demande ?

## Barres comparatives : 

```python
import pandas as pd
import plotly.express as px

def categorie_meteo(row):
    if row["snowfall"] > 0:
        return "Neige"
    elif row["rainfall"] > 0:
        return "Pluie"
    else:
        return "Sec"

df_active["meteo"] = df_active.apply(categorie_meteo, axis=1)

df_active["temp_bin"] = pd.cut(
    df_active["temperature"],
    bins=[-20, 0, 15, 40],
    labels=["Froid (<0°C)", "Doux (0-15°C)", "Chaud (>15°C)"]
)
resume = df_active.groupby(["temp_bin", "meteo"], observed=True)["rented_bike_count"].mean().reset_index()

color_discrete_map={
    "Sec": "#F4A460",      # sable/beige doré - évoque le soleil, temps sec
    "Pluie": "#4682B4",    # bleu acier - évoque l'eau, la pluie
    "Neige": "#B0C4DE",    # bleu très pâle/blanc bleuté - évoque la neige/glace
}

fig = px.bar(
    resume,
    x="temp_bin",
    y="rented_bike_count",
    color="meteo",
    barmode="group",
    color_discrete_map=color_discrete_map,
    labels={
        "temp_bin": "Tranche de température",
        "rented_bike_count": "Nombre moyen de vélos loués",
        "meteo": "Météo"
    },
    title="La pluie réduit fortement la demande à toute température ; la neige, plus rare, semble avoir un effet moindre en conditions douces"
)
fig.show()
```

## Commentaire :
### Chaud : la barre “Sec” est de loin la plus haute de tout le graphique beau temps chaud = pic de demande.
### Doux : “Sec” domine largement, “Pluie” chute très fort (la plus basse barre affichée), et “Neige” est étonnamment plus haute que Pluie dans cette tranche.
### Froid : les barres “Sec” et “Neige” sont proches l’une de l’autre, sans grand écart.
