#### InvoiceNo : Numéro de facture. Variable nominale ; nombre entier à 6 chiffres attribué de manière unique à chaque transaction. Si ce code commence par la lettre « c », il indique une annulation.
#### StockCode : Code du produit (article). Variable nominale ; nombre entier à 5 chiffres attribué de manière unique à chaque produit distinct.
#### Description : Nom du produit (article). Variable nominale.
#### Quantity : Quantité de chaque produit (article) par transaction. Variable numérique.
#### InvoiceDate : Date et heure de la facture. Variable numérique ; jour et heure auxquels chaque transaction a été générée.
#### UnitPrice : Prix unitaire. Variable numérique ; prix du produit à l'unité en livres sterling.
#### CustomerID : Numéro de client. Variable nominale ; nombre entier à 5 chiffres attribué de manière unique à chaque client.
#### Country : Nom du pays. Variable nominale ; nom du pays de résidence de chaque client.
